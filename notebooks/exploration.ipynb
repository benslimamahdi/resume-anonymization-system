# Resume Anonymization Exploration

This notebook demonstrates the usage of the refactored `resume-anonymization` library.
It uses the `anonymizer` package to detect and anonymize PII in text.

```python
# Ensure the project root is in the path
import sys
import os
sys.path.append(os.path.abspath('..'))

from anonymizer.detector import PIIDetector
from anonymizer.anonymizer import PIIAnonymizer
```

## 1. Setup Detector and Anonymizer

```python
detector = PIIDetector()
anonymizer = PIIAnonymizer()
```

## 2. Defined Input Text

```python
text = """
John Doe
1234 Elm Street, Springfield, IL
Phone: (555) 123-4567
Email: john.doe@example.com
LinkedIn: linkedin.com/in/johndoe123

Summary
Experienced Software Engineer.
"""
print("Original Text:")
print(text)
```

## 3. Detect PII

```python
results = detector.detect(text)
for res in results:
    print(f"Found {res.entity_type} score={res.score}")
```

## 4. Anonymize

```python
anonymized_text = anonymizer.anonymize(text, results)
print("Anonymized Text:")
print(anonymized_text)
```
